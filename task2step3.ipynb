```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset load karein
df = pd.read_csv('cleaned_sales_dataset.csv')

# 1. Correlation Heatmap (Numerical columns)
plt.figure(figsize=(6, 4))
corr = df[['Quantity', 'Unit_Price', 'Total_Sales']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Step 3: Correlation Heatmap')
plt.show()

# 2. Scatter Plot (Unit Price vs Total Sales by Category)
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Unit_Price', y='Total_Sales', hue='Category', alpha=0.7)
plt.title('Step 3: Unit Price vs Total Sales by Category')
plt.show()
```
